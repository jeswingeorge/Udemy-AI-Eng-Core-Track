Reinforcement Learning from Human Feedback (**RLHF**) was the critical process used to transition the raw GPT-3 model into the more capable and "chatty" **GPT-3.5** (and eventually ChatGPT). While GPT-3 was a powerful completion engine, it often produced toxic, unhelpful, or nonsensical outputs because it was only trained to predict the next word on the internet.

RLHF was implemented to "align" the model with human intent through a three-stage process:

### 1. Supervised Fine-Tuning (SFT)
The process began with human AI trainers writing out high-quality responses to specific prompts. For example, a trainer would write a helpful response to "Explain gravity to a five-year-old." These prompt-response pairs were used to fine-tune the base GPT-3 model. This taught the model the *format* of a helpful assistant.

### 2. The Reward Model (RM)
Once the model had a basic idea of how to respond, it was asked to generate multiple different answers for the same prompt. Human rankers then ordered these responses from best to worst based on accuracy, safety, and politeness. 
*   This data was used to train a separate **Reward Model**.
*   The Reward Model's job is to act as a "judge" that can mathematically score how much a human would like a specific response.



### 3. Reinforcement Learning (PPO)
In the final stage, the SFT model was placed in a feedback loop with the Reward Model. The model would generate a response, and the Reward Model would instantly provide a "score." 
*   Using an algorithm called **Proximal Policy Optimization (PPO)**, the model updated its internal weights to maximize its score.
*   If the model received a high score for being helpful and safe, it reinforced those behaviors. 

---

### Why RLHF was necessary
Before RLHF, large language models (LLMs) suffered from several "misalignment" issues that GPT-3.5 successfully addressed:

*   **Instruction Following:** GPT-3 might respond to a question with more questions. RLHF ensured GPT-3.5 actually followed the command.
*   **Hallucination Reduction:** By rewarding factual accuracy, the model became slightly less prone to making up "facts" (though it did not eliminate the problem entirely).
*   **Safety:** RLHF was used to bake in "guardrails," teaching the model to refuse requests for illegal acts or hate speech.

# [Transformers](https://youtube.com/shorts/6IgxN11wYPA?si=vC6hYtjzG-nGy4sE)

At its core, a **Transformer** is a type of neural network architecture that revolutionized how AI processes sequences of data—most notably human language. It was introduced by Google researchers in the 2017 paper *"Attention Is All You Need"* and serves as the foundation for models like GPT, Claude, and Gemini.

Before Transformers, AI models (like RNNs) read text one word at a time, from left to right. This made them slow and prone to "forgetting" the beginning of a long sentence by the time they reached the end. Transformers solved this by processing the entire sequence of data **all at once**.

---

## The Core Mechanisms

### 1. Self-Attention
This is the "secret sauce." Instead of looking at words in isolation, Self-Attention allows the model to look at every other word in a sentence to understand context. 
*   **Example:** In the sentence *"The animal didn't cross the street because **it** was too tired,"* the model uses attention to link "**it**" to "**animal**." If the sentence ended in *"because **it** was too wide,"* the model would shift its attention to link "**it**" to "**street**."



### 2. Positional Encoding
Since Transformers process all words simultaneously, they don't naturally know the order of words. Positional Encoding adds a mathematical "tag" to each word to tell the model exactly where it sits in the sentence (e.g., that "The" is at position 1 and "Cat" is at position 2).

### 3. Parallelization
Unlike older models that had to wait for one word to finish before starting the next, Transformers can be trained on massive GPUs in parallel. This leap in efficiency is what allowed researchers to train models on the entire public internet.

---

## The Architecture: Encoder and Decoder
The original Transformer consists of two main parts:

*   **The Encoder:** It "reads" and understands the input text, converting it into a complex numerical map (embeddings) that captures meaning and context.
*   **The Decoder:** It takes that map and "writes" the output, predicting the next word in a sequence one by one.



---

## Why They Matter
Transformers shifted the AI landscape from "narrow" tasks (like simple translation) to **Generative AI**. Because they can understand deep relationships within data, they aren't just limited to text; they are now used for:
*   **Computer Vision:** Analyzing images as sequences of pixels.
*   **Protein Folding:** Predicting biological structures in medicine.
*   **Music & Audio:** Generating waveforms and compositions.

## Parameters

In the context of Large Language Models (LLMs), **parameters** are essentially the "knowledge" of the model. They are the internal variables that the model learns and adjusts during the training process.

Think of an LLM as a massive mathematical function. If a simple linear equation is $y = mx + c$, then $m$ and $c$ are the parameters. In an LLM, there are billions of these variables.

### What do they represent?
Parameters are primarily composed of **weights** and **biases** within the Transformer's neural network layers. 
*   **Weights:** These determine the strength of the connection between "neurons." They decide how much importance one piece of information (like a word) has relative to another in a specific context.
*   **Biases:** These are extra values added to the weights to help the model better fit the data, allowing for more flexible decision-making.



### How do they work?
When you see a model described as "70B" (like Llama-3 70B), it means the model has **70 billion parameters**. 
1.  **During Training:** The model starts with random parameters. As it reads text, it makes predictions. If a prediction is wrong, it uses an algorithm (backpropagation) to slightly nudge billions of parameters so that it is more likely to get the answer right next time.
2.  **During Inference (Usage):** When you type a prompt, those billions of parameters are used as fixed numerical "filters" that the input text passes through to calculate the most probable next word.

### Why do more parameters matter?
Generally, a higher parameter count allows the model to:
*   **Capture Nuance:** Understand complex patterns, sarcasm, and deep technical logic.
*   **Store Knowledge:** Retain more factual information from its training data.
*   **Reasoning:** Handle multi-step instructions more effectively.

However, more parameters also mean the model requires more **VRAM** and computational power to run, which is why there is a constant push to make smaller models (like "7B" or "8B" versions) more efficient through techniques like quantization.

### Wait, wha??

We just told you!

What's going on??

Here's the thing: every call to an LLM is completely STATELESS. It's a totally new call, every single time. As AI engineers, it's OUR JOB to devise techniques to give the impression that the LLM has a "memory".

```python
messages = [
    {"role": "system", "content": "You are a helpful assistant"},
    {"role": "user", "content": "Hi! I'm Ed!"},
    {"role": "assistant", "content": "Hi Ed! How can I assist you today?"},
    {"role": "user", "content": "What's my name?"}
    ]
```

```python
response = openai.chat.completions.create(model="gpt-4.1-mini", messages=messages)
response.choices[0].message.content
```

## To recap

With apologies if this is obvious to you - but it's still good to reinforce:

1. Every call to an LLM is stateless
2. We pass in the entire conversation so far in the input prompt, every time
3. This gives the illusion that the LLM has memory - it apparently keeps the context of the conversation
4. But this is a trick; it's a by-product of providing the entire conversation, every time
5. An LLM just predicts the most likely next tokens in the sequence; if that sequence contains "My name is Ed" and later "What's my name?" then it will predict.. Ed!

The ChatGPT product uses exactly this trick - every time you send a message, it's the entire conversation that gets passed in.

"Does that mean we have to pay extra each time for all the conversation so far"

For sure it does. And that's what we WANT. We want the LLM to predict the next tokens in the sequence, looking back on the entire conversation. We want that compute to happen, so we need to pay the electricity bill for it!
